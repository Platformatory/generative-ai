### RAG Fundamentals & Databricks AI Search

- **3.1 Context engineering vs. prompt engineering** - two ways to adapt a model (change what it sees vs. change its weights); why context engineering is the primary discipline
- **3.2 RAG pattern** - indexing phase (documents → chunk → embed → vector DB) vs. query phase (question → embed → similarity search → grounded answer)
- **3.3 Why retrieval agents exist** - knowledge cutoffs, hallucination, missing private context
- **3.4 Document processing pipeline** - `ai_parse_document` (PDF → structured JSON) → `ai_classify`/`ai_extract` (categorize/structure) → `ai_prep_search` (chunk)
- **3.5 Chunking strategy considerations** - chunk size vs. embedding model limits, overlap (10–20%), granularity trade-off, "lost in the middle" phenomenon
- **3.6 chunk_to_embed vs. chunk_to_retrieve** - context-enriched text for findability vs. clean raw text for the LLM
- **3.7 Embeddings & cosine similarity** - how vectors capture meaning; how AI Search ranks results (ANN)
- **3.8 Creating/syncing an AI Search index** - key parameters (`endpoint_name`, `source_table_name`, `embedding_model_endpoint_name`, `pipeline_type`); TRIGGERED vs. CONTINUOUS sync; Change Data Feed
- **3.9 Managed RAG (Knowledge Assistant) vs. Custom RAG (Agent Framework + AI Search)** - setup time, customizability, when to use each
- **3.10 Unity Catalog governance across the retrieval pipeline** - UC privileges on volumes/tables/indexes, endpoint ACLs

### Multi-Agent Systems, Agent Bricks & Genie Integration


- **5.1 Single-agent vs. multi-agent** - decision factors: tool count (~8–10 threshold), domain complexity, instruction length, failure isolation, team ownership, latency
- **5.2 Supervisor-worker (router-delegate) pattern** - supervisor routes/delegates to specialized workers (Data Agent, Research Agent, General Agent), synthesizes response
- **5.3 OpenAI Agents SDK orchestration patterns** - LLM-driven (model decides) vs. code-driven (deterministic flow); handoffs
- **5.4 `@function_tool` decorator** - turning a Python function into an agent tool for rapid prototyping (name/type hints/docstring → tool schema)
- **5.5 Agent Bricks** - no-code platform for building/optimizing/governing production agents; supported types: Knowledge Assistant, Supervisor Agent, Document Intelligence, Custom Agents
- **5.6 Agent Bricks development lifecycle** - specify → configure/evaluate on your data → continuously improve (built on UC, Model Serving, and Agent Evaluation)
- **5.7 Genie Space vs. Genie Agent** - curated NL interface to governed data; used standalone or as a specialized worker in a multi-agent system
- **5.8 Integrating Genie with agents** - as a subagent under a supervisor
- **5.9 Code-first (OpenAI SDK, LangChain, LangGraph, DSPy) vs. no-code (Agent Bricks)** - when to reach for each
